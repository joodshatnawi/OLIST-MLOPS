```python
import pandas as pd 
from sqlalchemy import create_engine
import numpy as np

engine = create_engine("postgresql+psycopg2://olist:olist123@localhost:5432/olist")
```

```python
try:
    with engine.connect() as conn:
        print("Database connection successful!")
except Exception as e:
    print("Database connection failed!")
    print(e)
```

```python
table_names = [
    "customers",
    "geolocation",
    "orders",
    "order_items",
    "order_payments",
    "products",
    "sellers",
    "orders_reviews",
    "product_category_name_translation"
]

tables = {}

for table_name in table_names:
    tables[table_name] = pd.read_sql(
        f'SELECT * FROM "{table_name}"',
        engine
    )

    print(f"{table_name}: {tables[table_name].shape}")
```

```python
for table_name, df in tables.items():
    print("=" * 70)
    print(table_name)
    print("=" * 70)
    print(df.columns.tolist())
    print()
```

```python
for table_name, df in tables.items():
    print("=" * 70)
    print(table_name)
    print("=" * 70)
    display(df.head(3))
    print()
```

```python
key_columns = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id"
}

for table_name, key in key_columns.items():
    df = tables[table_name]

    duplicate_count = df[key].duplicated().sum()

    print(
        f"{table_name}.{key} -> "
        f"{duplicate_count} duplicate values"
    )
```

```python
order_items = tables["order_items"].copy()

items_agg = (
    order_items
    .groupby("order_id")
    .agg(
        item_count=("order_item_id", "count"),
        total_items_price=("price", "sum"),
        total_freight_value=("freight_value", "sum"),
        unique_products=("product_id", "nunique"),
        unique_sellers=("seller_id", "nunique")
    )
    .reset_index()
)

print("Aggregated order_items shape:", items_agg.shape)

display(items_agg.head())
```

```python
print("Unique orders in order_items:",
      order_items["order_id"].nunique())

print("Rows after aggregation:",
      len(items_agg))

print(
    "One row per order:",
    items_agg["order_id"].is_unique
)
```

```python
order_payments = tables["order_payments"].copy()

payments_agg = (
    order_payments
    .groupby("order_id")
    .agg(
        payment_count=("payment_sequential", "count"),
        total_payment_value=("payment_value", "sum"),
        payment_types=("payment_type", "nunique"),
        max_installments=("payment_installments", "max")
    )
    .reset_index()
)

print("Aggregated order_payments shape:", payments_agg.shape)

display(payments_agg.head())
```

```python
print("Unique orders in payments:",
      order_payments["order_id"].nunique())

print("Rows after aggregation:",
      len(payments_agg))

print(
    "One row per order:",
    payments_agg["order_id"].is_unique
)
```

```python
orders = tables["orders"].copy()

print("Orders shape:", orders.shape)
print("Unique order IDs:", orders["order_id"].nunique())
```

```python
ml_table = orders.merge(
    items_agg,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("After joining items:", ml_table.shape)
```

```python
ml_table = ml_table.merge(
    payments_agg,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("After joining payments:", ml_table.shape)
```

```python
customers = tables["customers"].copy()

ml_table = ml_table.merge(
    customers,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

print("After joining customers:", ml_table.shape)
```

```python
reviews = tables["orders_reviews"].copy()

review_agg = (
    reviews
    .groupby("order_id")
    .agg(
        review_score=("review_score", "mean")
    )
    .reset_index()
)

ml_table = ml_table.merge(
    review_agg,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("After joining reviews:", ml_table.shape)
```

```python
print("Final rows:", len(ml_table))
print("Unique orders:", ml_table["order_id"].nunique())
print("Order ID unique:", ml_table["order_id"].is_unique)

assert ml_table["order_id"].is_unique

print("\n✅ Final ML table contains exactly one row per order.")
```

```python
print("Final shape:", ml_table.shape)

display(ml_table.head())

print("\nColumns:")
print(ml_table.columns.tolist())
```

```python
from pathlib import Path

artifacts_dir = Path("../artifacts")
artifacts_dir.mkdir(parents=True, exist_ok=True)

print("Artifacts directory ready:", artifacts_dir.resolve())
```

```python
output_path = artifacts_dir / "ml_table.csv"

# Save the dataframe in chunks to reduce memory usage
ml_table.to_csv(
    output_path,
    index=False,
    chunksize=5000
)

print("Artifact saved successfully:")
print(output_path.resolve())
```
